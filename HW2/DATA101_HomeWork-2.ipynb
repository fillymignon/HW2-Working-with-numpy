```python
import numpy as np
import random

names = ['Francis', 'Jack', 'Jacquelyn', 'Samuel', 'Esther', 'Leo', 'Catherine', 'Norris', 'Bruce', 'Anna']
grades = []
scores = np.zeros((10, 5))
assesments = ['homework', 'project' , 'exam 1', 'exam 2' , 'final']
for x in range(10):
    print(f"Enter 5 grades (0 to 100) for {names[x]} (student {x+1})")
    for y in range(5):
        grades = float(input(f"Enter their grade for the {assesments[y]}"))
        scores[x,y] = grades
weight = [0.15, 0.15, .20, .20, .30]
final = scores@weight
for z in range(len(final)):
    mark = final[z]

    if mark >= 90:
        print(f"{names[z]} received a grade of A")
    elif mark >= 80:
        print(f"{names[z]} received a grade of B")
    elif mark >= 70:
        print(f"{names[z]} received a grade of C")
    elif mark >= 60:
        print(f"{names[z]} received a grade of D")
    else:
        print(f"{names[z]} received a grade of F")
```
